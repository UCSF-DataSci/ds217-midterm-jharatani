## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import utilities
from q3_data_utils import load_data, summarize_by_group

df = load_data('output/q6_transformed_data.csv')
print(f"Loaded {len(df)} patients")

# Prewritten visualization function for grouped analysis
def plot_group_comparison(data, x_col, y_col, title):
    """
    Create a bar chart comparing groups.
    
    Args:
        data: DataFrame with grouped data
        x_col: Column name for x-axis (groups)
        y_col: Column name for y-axis (values)
        title: Chart title
    """
    plt.figure(figsize=(10, 6))
    data.plot(x=x_col, y=y_col, kind='bar')
    plt.title(title)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## Part 1: Basic Groupby (5 points)

1. Group by 'site' and calculate mean age, BMI, and blood pressure
2. Group by 'intervention_group' and count patients
3. Use the `summarize_by_group()` utility to get overall statistics by site

```python
# TODO: Group by site

print(df.groupby('site')[['age','bmi','systolic_bp','diastolic_bp']].mean())

```

```python
# TODO: Count by intervention group

intervention_counts = df['intervention_group'].value_counts()
print("\nIntervention Group Counts:")
print(intervention_counts)
```

**Note:** The `summarize_by_group()` function has an optional `agg_dict` parameter for custom aggregations. If you don't specify it, it will use `.describe()` on numeric columns. You can use `agg_dict={'age': ['mean', 'std'], 'bmi': 'mean'}` for custom aggregations.


```python
# TODO: Use summarize_by_group utility

grouped_summary = summarize_by_group(df, 'site', agg_dict= {"age": ["mean", "median", "min", "max","std"],
              "bmi": ["mean", "median", "min", "max","std"],
              "systolic_bp": ["mean", "median", "min", "max","std"],
              "diastolic_bp": ["mean", "median", "min", "max","std"]}
)

print("\nGrouped Summary by Site:")
display(grouped_summary)
```

## Part 2: Multiple Aggregations (5 points)

Group by 'site' and apply multiple aggregations:
- age: mean, std, min, max
- bmi: mean, std
- systolic_bp: mean, median

Display the results in a well-formatted table.

```python
# TODO: Multiple aggregations

agg_results = df.groupby('site').agg({
    'age': ['mean', 'std', 'min', 'max'],
    'bmi': ['mean', 'std'],
    'systolic_bp': ['mean', 'median'],
})

print("\nMultiple Aggregations by Site:")
agg_results.columns = ['age (mean)', 'age (std)', 'age (min)', 'age (max)',
                       'bmi (mean)', 'bmi (std)',
                       'systolic_bp (mean)', 'systolic_bp (median)']

display(agg_results)
```

## Part 3: Comparative Analysis (5 points)

Compare intervention groups:
1. Calculate mean outcome_cvd rate by intervention_group
2. Calculate mean adherence_pct by intervention_group
3. Create a cross-tabulation of intervention_group vs dropout status
4. Visualize the comparison with a bar plot

```python
# TODO: Intervention group comparisons

#The rate of outcome_cvd (Yes) by intervention group
outcome_cvd_rate_by_intervention = (df.groupby("intervention_group")["outcome_cvd"].apply(lambda x: (x == "Yes").mean())
.reset_index(name="outcome_cvd_mean_rate")
)
print("\nOutcome_cvd yes rate by intervention")
print(outcome_cvd_rate_by_intervention)

mean_adherence_by_intervention = (df.groupby("intervention_group")["adherence_pct"].mean()
.reset_index(name="mean_adherence_pct")
)
print("\nMean adherence percentage by intervention")
print(mean_adherence_by_intervention)

#Cross-tabulation of intervention group and dropout
crosstab = pd.crosstab(df["intervention_group"], df["dropout"])
print("\nCrosstab of intervention group and dropout")
display(crosstab)

```

```python
# TODO: Visualization
#Visualization of dropout by intervention group
crosstab.plot(kind='bar', figsize=(8, 5), title='Dropout Rate by Intervention Group')
plt.ylabel('Rate of Dropout')
plt.xlabel('Intervention Group')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Create and save:

```python
# TODO: Save summary outputs

grouped_summary.to_csv('output/q7_site_summary.csv', index=False)

outcome_cvd_rate_by_intervention.to_csv('output/q7_intervention_comparison.csv', index=False)
mean_adherence_by_intervention.to_csv('output/q7_intervention_comparison.csv', mode = 'a', header=True, index=False)
with open("output/q7_intervention_comparison.csv", "a") as f:  
    f.write("\n=== Dropout Counts by Intervention Group ===\n")
    crosstab.to_csv(f)

txt_report_key_findings = """
Key Findings from Aggregation Analysis:
1. Average age and BMI do not vary significantly across different sites, indicating a good choice of sites and consistency for a more controlled study.
2. The Control group shows a higher dropout rate compared to Treatment groups A and B.
3. The rate of cardiovascular outcomes ('outcome_cvd') is notably higher in Control group.
4. Mean adherence percentage is highest in intervention group 'B', suggesting better compliance in this treatment group.
"""
with open('output/q7_key_findings.txt', 'w') as f:
    f.write(txt_report_key_findings)

    print("\nKey findings report saved to output/q7_key_findings.txt")
```
